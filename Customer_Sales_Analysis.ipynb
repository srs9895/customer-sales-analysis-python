```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df["Order_Date"].dtype
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.isnull().sum()
```

```python
total_sales = df["Sales"].sum()
print("total_sales:",total_sales)
```

```python
Avergage_sales = df["Sales"].mean()
print("Avergage_sales:",Avergage_sales)
```

```python
Highest_sales = df["Sales"].max()
print("Highest_sales:",Highest_sales)
```

```python
Lowest_sales = df["Sales"].min()
print("Lowest_sales:",Lowest_sales)
```

```python
Total_quantity = df["Quantity"].sum()
print("Total_quantity:",Total_quantity)
```

```python
Product_Sales = df.groupby("Product")["Sales"].sum()
print("Product_Sales:\n",Product_Sales)
```

```python
Product_Sales = Product_Sales.sort_values(ascending=False)
print("Product_Sales:\n",Product_Sales)
```

```python
best_product = Product_Sales.idxmax()
print("best_product:",best_product)
```

```python
Product_Sales.head(5)
```

```python
category_sales = df.groupby("Category")["Sales"].sum()
print("category_sales:\n",category_sales)
```

```python
category_sales.sort_values(ascending=False)
```

```python
best_category = category_sales.idxmax()
print("best_category:",best_category)
```

```python
City_Sales = df.groupby("City")["Sales"].sum()
print("City_Sales:\n",City_Sales)
```

```python
City_Sales.sort_values(ascending=False)
```

```python
best_city = City_Sales.idxmax()
print("best_city:",best_city)
```

```python
City_Sales.head(5)
```

```python
Customername_Sales = df.groupby("Sales")["Customer_ID"].sum()
print(Customername_Sales.head(5))
```

```python
Top_customer = Customername_Sales.idxmax()
print("Top_customer:",Top_customer)
```

```python
customer_sales = df.groupby("Customer_ID")["Sales"].sum()

print(customer_sales.head())
```

```python
top_customer = customer_sales.idxmax()

print("Top Customer:", top_customer)
```

```python
customer_sales.sort_values(ascending=False).head(10)
```

```python
unique_customers = df["Customer_ID"].nunique()

print("Unique Customers:", unique_customers)
```

```python
df["Gender"].value_counts()
```

```python
Sales_Gender = df.groupby("Gender")["Sales"].sum()
print("Sales By Gender",Sales_Gender)
```

```python
df["Payment_Method"].value_counts()
```

```python
Sales_By_Payment_method = df.groupby("Payment_Method")["Sales"].sum()
print("sales by payment method:\n",Sales_By_Payment_method)
```

```python
df["Payment_Method"].value_counts().idxmax()
```

```python
df["Age"].mean()
```

```python
df["Age"].min()
```

```python
df["Month"] = df["Order_Date"].dt.month
```

```python
df[["Order_Date", "Month"]].head()
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

print(monthly_sales)
```

```python
best_month = monthly_sales.idxmax()

print("Best Month:", best_month)
```

```python
import matplotlib.pyplot as plt
```

```python
Product_Sales.plot(kind="bar")
plt.xlabel("Product")
plt.ylabel("Sales")
plt.title("Product Sales")
plt.show()
```

```python
City_Sales.plot(kind="bar")
plt.title("Sales By City")
plt.xlabel("City")
plt.ylabel("Sales")
plt.show()
```

```python
category_sales.plot(kind="pie",autopct="%1.1f%%")
plt.title("Sales By Category")

plt.ylabel("")
plt.show()
```

```python
monthly_sales.plot(kind="line",marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```
